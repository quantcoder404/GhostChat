```python
import pandas as pd
import numpy as np
```

```python
labels = ['a', 'b', 'c']
my_list = [10,11,12]
arr = np.array([10,20,30])
dicts = {0:10, 1:15, False:30, "a":45}     # 1)zero and false comflict both act as 0. 2)if 0 label not exist then false act as flase label
```

```python
pd.Series(labels)
```

```python
pd.Series(arr)
```

```python
pd.Series(my_list)
```

```python
pd.Series(my_list, index=labels)
# same
# pd.Series(my_list, labels)
```

```python
pd.Series(dicts)
# false act as 0 label
```

```python
dicts = { 1:15, False:30, "a":45}
pd.Series(dicts)
# no zero label exist so false act as false label
```
